```python
# 1. Define our automated hardware referee
def symbolic_rule_checker(prompt, generated_code):
    # Rule A: In Verilog, every single statement MUST end with a semicolon ';'
    if not generated_code.strip().endswith(";"):
        return "❌ FAIL: Critical Syntax Error! Missing a semicolon at the end."

    # Rule B: If the prompt asks for an 'AND' gate, the code must use the electrical symbol '&'
    if "AND gate" in prompt and "&" not in generated_code:
        return "❌ FAIL: Logical Error! AND gate requires the '&' operator."

    return "✅ PASS: Hardware code is logically and physically valid!"

# 2. Let's test a piece of code that an AI might generate
mock_prompt = "Design a 2-input AND gate"
mock_ai_output = "assign out = a & b;"  # Notice I purposely left out the semicolon here!

# 3. Run it through the referee
result = symbolic_rule_checker(mock_prompt, mock_ai_output)
print("Referee Report:", result)
```

```python
# 1. Start with the broken AI code again
ai_output = "assign out = a & b"
print("Original Broken AI Code:", ai_output)
print("-" * 40)

# 2. The Autonomous Loop (Like a mini AlphaChip)
attempts = 1
while True:
    # Check the current code with our referee
    report = symbolic_rule_checker("Design a 2-input AND gate", ai_output)
    print(f"Attempt #{attempts} Report: {report}")

    # If it passes, we stop!
    if "✅ PASS" in report:
        print(f"\n🎉 SUCCESS! The chip blueprint is flawless after {attempts} attempts.")
        print("Final Saved Blueprint:", ai_output)
        break

    # If it fails the semicolon rule, the AI automatically applies a patch
    if "Missing a semicolon" in report:
        print("🤖 System Feedback: Automatically adding missing hardware syntax...")
        ai_output = ai_output + ";"  # Self-correction happens here

    attempts += 1
```

```python
# 1. Define an advanced CoT referee that inspects both plan and code
def cot_hardware_referee(ai_generated_text):
    # Split the output to isolate the Plan and the Code
    try:
        plan_part = ai_generated_text.split("Plan:")[-1].split("Code:")[0].strip()
        code_part = ai_generated_text.split("Code:")[-1].strip().split("\n")[0].strip()
    except Exception:
        return "❌ FAIL: Parsing Error! Could not cleanly separate Plan and Code."

    print(f"📝 Extracted Plan: {plan_part}")
    print(f"💻 Extracted Code: {code_part}")

    # Rule 1: Check if the plan is detailed enough
    if len(plan_part) < 10:
        return "❌ FAIL: Plan is too brief or missing architectural details."

    # Rule 2: Check if code ends with a semicolon
    if not code_part.endswith(";"):
        return "❌ FAIL: Syntax Error! Missing a semicolon at the end of the code."

    # Rule 3: Check sequential vs combinatorial logic rules
    if "Flip-Flop" in ai_generated_text and "always" not in code_part:
        return "❌ FAIL: Logic Error! Sequential memory requires an 'always' clock block."

    return "✅ PASS: Both Microarchitecture Plan and Verilog Code Verified!"

# 2. Test our new CoT referee against the output we just generated
print("--- RUNNING CoT REFEREE EVALUATION ---")
verdict = cot_hardware_referee(cot_output)
print(f"Referee Verdict: {verdict}")
```

```python
import subprocess
import os
from transformers import pipeline

# Initialize AI brain
ai_brain = pipeline("text-generation", model="distilgpt2")

hardware_prompt = "Prompt: Design a Verilog module for a 2-input AND gate\nPlan: Define module with inputs a, b and output out, then assign out = a & b.\nCode: module and_gate(input a, input b, output out); assign out = a & b; endmodule"

print("⚡ INITIALIZING CLOSED-LOOP SELF-CORRECTION PIPELINE...\n" + "="*60)

max_retries = 2
success = False

for attempt in range(1, max_retries + 1):
    print(f"🔄 Generation & Compilation Attempt #{attempt}...")

    # Generate response
    response = ai_brain(
        hardware_prompt,
        max_length=80,
        repetition_penalty=1.3,
        do_sample=True,
        temperature=0.2
    )[0]['generated_text']

    # Extract code safely
    try:
        code = response.split("Code:")[-1].strip()
    except Exception:
        code = "module and_gate(input a, input b, output out); assign out = a & b; endmodule"

    if "endmodule" not in code:
        code += " endmodule"

    print(f"   💻 Generated Code:\n{code}")

    # Write to file and compile
    filename = "ai_design.v"
    with open(filename, "w") as f:
        f.write(code)

    compilation = subprocess.run(["iverilog", "-o", "sim.out", filename], capture_output=True, text=True)

    if compilation.returncode == 0:
        print("✅ COMPILATION SUCCESSFUL! The self-healing loop verified the hardware.")
        success = True
        break
    else:
        print(f"❌ Compiler Error Caught: {compilation.stderr.strip()}")
        # Feed error back into the prompt for the next attempt
        hardware_prompt += f"\nPrevious attempt failed with error: {compilation.stderr.strip()}\nFix the syntax and rewrite Code:"
        print("🛠️ Feeding compiler feedback back into AI model for automatic repair...\n" + "-"*50)

if not success:
    print("⚠️ Max retries reached. Human-in-the-loop review required.")

# Clean up
if os.path.exists("sim.out"):
    os.remove("sim.out")
```
