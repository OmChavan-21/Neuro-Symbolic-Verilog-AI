```python
# 1. Define the human prompt and the correct chip code
prompt = "Design a basic 2-input AND gate chip"
verilog_code = "assign out = a & b;"

# 2. Print them out together so the AI can see the training pair
print("--- AI TRAINING PAIR ---")
print("User Request:", prompt)
print("Correct Hardware Code:", verilog_code)
```

```python
# 1. Create a miniature dataset (a list of multiple chip designs)
dataset = [
    {"prompt": "Design a 2-input AND gate", "code": "assign out = a & b;"},
    {"prompt": "Design a 2-input OR gate", "code": "assign out = a | b;"},
    {"prompt": "Design a NOT gate inverter", "code": "assign out = ~a;"}
]

# 2. Check how many examples we have
dataset_size = len(dataset)
print(f"Success! You just built a hardware dataset with {dataset_size} examples.")

# 3. Look at one specific example inside the dataset
print("The third example in your dataset is:", dataset[2])
```

```python
# 1. Define our automated hardware referee
def symbolic_rule_checker(prompt, generated_code):
    # Rule A: In Verilog, every single statement MUST end with a semicolon ';'
    if not generated_code.strip().endswith(";"):
        return "❌ FAIL: Critical Syntax Error! Missing a semicolon at the end."

    # Rule B: If the prompt asks for an 'AND' gate, the code must use the electrical symbol '&'
    if "AND gate" in prompt and "&" not in generated_code:
        return "❌ FAIL: Logical Error! AND gate requires the '&' operator."

    return "✅ PASS: Hardware code is logically and physically valid!"

# 2. Let's test a piece of code that an AI might generate
mock_prompt = "Design a 2-input AND gate"
mock_ai_output = "assign out = a & b;"  # Notice I purposely left out the semicolon here!

# 3. Run it through the referee
result = symbolic_rule_checker(mock_prompt, mock_ai_output)
print("Referee Report:", result)
```

```python
# 1. Start with the broken AI code again
ai_output = "assign out = a & b"
print("Original Broken AI Code:", ai_output)
print("-" * 40)

# 2. The Autonomous Loop (Like a mini AlphaChip)
attempts = 1
while True:
    # Check the current code with our referee
    report = symbolic_rule_checker("Design a 2-input AND gate", ai_output)
    print(f"Attempt #{attempts} Report: {report}")

    # If it passes, we stop!
    if "✅ PASS" in report:
        print(f"\n🎉 SUCCESS! The chip blueprint is flawless after {attempts} attempts.")
        print("Final Saved Blueprint:", ai_output)
        break

    # If it fails the semicolon rule, the AI automatically applies a patch
    if "Missing a semicolon" in report:
        print("🤖 System Feedback: Automatically adding missing hardware syntax...")
        ai_output = ai_output + ";"  # Self-correction happens here

    attempts += 1
```

```python
# 1. Download the AI "Transformers" tool library
import transformers
from transformers import pipeline

print("Downloading a real AI model brain into your notebook... Please wait a moment...")

# 2. Load a lightweight text-generation model brain
ai_brain = pipeline("text-generation", model="distilgpt2")

# 3. Test the real AI model's creativity
prompt = "The most important rule in microchip design is"
ai_prediction = ai_brain(prompt, max_length=30, num_return_sequences=1)

print("\n--- REAL AI RESPONSE ---")
print(ai_prediction[0]['generated_text'])
```

```python
# 1. Create a prompt that teaches the AI by showing it examples first
training_prompt = """
Prompt: Design a 2-input AND gate
Code: assign out = a & b;

Prompt: Design a 2-input OR gate
Code: assign out = a | b;

Prompt: Design a NOT gate inverter
Code:"""

# 2. Feed this structured training prompt to your AI brain
print("The AI is analyzing the patterns in your dataset...")
ai_hardware_prediction = ai_brain(training_prompt, max_length=60, num_return_sequences=1)

print("\n--- SPECIALIZED AI HARDWARE RESPONSE ---")
print(ai_hardware_prediction[0]['generated_text'])
```

```python
# 1. Upgrade our logic referee to handle NOT gates
def advanced_hardware_referee(prompt, code):
    if not code.strip().endswith(";"):
        return "❌ FAIL: Syntax Error! Missing a semicolon at the end."

    # Check if a NOT gate request contains the electrical inversion symbol '~'
    if "NOT gate" in prompt and "~" not in code:
        return "❌ FAIL: Logic Error! A NOT gate (inverter) requires the '~' operator, not '&'."

    return "✅ PASS: Chip logic verified!"

# 2. Run the AI brain again, but add anti-loop parameters
print("Re-running the model with strict anti-repetition rules...")
fixed_output = ai_brain(
    training_prompt,
    max_length=62,
    repetition_penalty=1.3,  # 👈 This forces the AI to stop repeating itself
    do_sample=True,          # 👈 Allows the model to explore different options
    temperature=0.3          # 👈 Keeps it focused instead of wildly creative
)

# 3. Print out the raw result to see if the loop stopped
ai_text = fixed_output[0]['generated_text']
print("\n--- ANTI-LOOP AI GENERATION ---")
print(ai_text)

# 4. Pull out just the code line the AI attempted for the NOT gate and test it
# (We look for what came right after the last 'Code:')
try:
    ai_final_line = ai_text.split("Code:")[-1].strip().split("\n")[0]
    print("\n--- REFEREE EVALUATION ---")
    print(f"AI attempted code: {ai_final_line}")
    print(f"Referee Verdict: {advanced_hardware_referee('Design a NOT gate inverter', ai_final_line)}")
except Exception as e:
    print("Could not isolate final line for grading.")
```
