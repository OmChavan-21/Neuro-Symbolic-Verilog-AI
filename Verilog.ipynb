```python
# 1. Create a miniature dataset (a list of multiple chip designs)
dataset = [
    {"prompt": "Design a 2-input AND gate", "code": "assign out = a & b;"},
    {"prompt": "Design a 2-input OR gate", "code": "assign out = a | b;"},
    {"prompt": "Design a NOT gate inverter", "code": "assign out = ~a;"}
]

# 2. Check how many examples we have
dataset_size = len(dataset)
print(f"Success! You just built a hardware dataset with {dataset_size} examples.")

# 3. Look at one specific example inside the dataset
print("The third example in your dataset is:", dataset[2])
```

```python
# 1. Define our automated hardware referee
def symbolic_rule_checker(prompt, generated_code):
    # Rule A: In Verilog, every single statement MUST end with a semicolon ';'
    if not generated_code.strip().endswith(";"):
        return "❌ FAIL: Critical Syntax Error! Missing a semicolon at the end."

    # Rule B: If the prompt asks for an 'AND' gate, the code must use the electrical symbol '&'
    if "AND gate" in prompt and "&" not in generated_code:
        return "❌ FAIL: Logical Error! AND gate requires the '&' operator."

    return "✅ PASS: Hardware code is logically and physically valid!"

# 2. Let's test a piece of code that an AI might generate
mock_prompt = "Design a 2-input AND gate"
mock_ai_output = "assign out = a & b;"  # Notice I purposely left out the semicolon here!

# 3. Run it through the referee
result = symbolic_rule_checker(mock_prompt, mock_ai_output)
print("Referee Report:", result)
```

```python
# 1. Start with the broken AI code again
ai_output = "assign out = a & b"
print("Original Broken AI Code:", ai_output)
print("-" * 40)

# 2. The Autonomous Loop (Like a mini AlphaChip)
attempts = 1
while True:
    # Check the current code with our referee
    report = symbolic_rule_checker("Design a 2-input AND gate", ai_output)
    print(f"Attempt #{attempts} Report: {report}")

    # If it passes, we stop!
    if "✅ PASS" in report:
        print(f"\n🎉 SUCCESS! The chip blueprint is flawless after {attempts} attempts.")
        print("Final Saved Blueprint:", ai_output)
        break

    # If it fails the semicolon rule, the AI automatically applies a patch
    if "Missing a semicolon" in report:
        print("🤖 System Feedback: Automatically adding missing hardware syntax...")
        ai_output = ai_output + ";"  # Self-correction happens here

    attempts += 1
```

```python
# 1. Create a prompt that teaches the AI by showing it examples first
training_prompt = """
Prompt: Design a 2-input AND gate
Code: assign out = a & b;

Prompt: Design a 2-input OR gate
Code: assign out = a | b;

Prompt: Design a NOT gate inverter
Code:"""

# 2. Feed this structured training prompt to your AI brain
print("The AI is analyzing the patterns in your dataset...")
ai_hardware_prediction = ai_brain(training_prompt, max_length=60, num_return_sequences=1)

print("\n--- SPECIALIZED AI HARDWARE RESPONSE ---")
print(ai_hardware_prediction[0]['generated_text'])
```

```python
# 1. Upgrade our logic referee to handle NOT gates
def advanced_hardware_referee(prompt, code):
    if not code.strip().endswith(";"):
        return "❌ FAIL: Syntax Error! Missing a semicolon at the end."

    # Check if a NOT gate request contains the electrical inversion symbol '~'
    if "NOT gate" in prompt and "~" not in code:
        return "❌ FAIL: Logic Error! A NOT gate (inverter) requires the '~' operator, not '&'."

    return "✅ PASS: Chip logic verified!"

# 2. Run the AI brain again, but add anti-loop parameters
print("Re-running the model with strict anti-repetition rules...")
fixed_output = ai_brain(
    training_prompt,
    max_length=62,
    repetition_penalty=1.3,  # 👈 This forces the AI to stop repeating itself
    do_sample=True,          # 👈 Allows the model to explore different options
    temperature=0.3          # 👈 Keeps it focused instead of wildly creative
)

# 3. Print out the raw result to see if the loop stopped
ai_text = fixed_output[0]['generated_text']
print("\n--- ANTI-LOOP AI GENERATION ---")
print(ai_text)

# 4. Pull out just the code line the AI attempted for the NOT gate and test it
# (We look for what came right after the last 'Code:')
try:
    ai_final_line = ai_text.split("Code:")[-1].strip().split("\n")[0]
    print("\n--- REFEREE EVALUATION ---")
    print(f"AI attempted code: {ai_final_line}")
    print(f"Referee Verdict: {advanced_hardware_referee('Design a NOT gate inverter', ai_final_line)}")
except Exception as e:
    print("Could not isolate final line for grading.")
```

```python
# 1. An expanded, industrial-grade micro-dataset
advanced_dataset = [
    {
        "prompt": "Design a 2-to-1 Multiplexer (MUX)",
        "code": "assign out = sel ? b : a;"
    },
    {
        "prompt": "Design a 1-bit Half Adder",
        "code": "assign sum = a ^ b; assign carry = a & b;"
    },
    {
        "prompt": "Design a D Flip-Flop memory cell",
        "code": "always @(posedge clk) q <= d;"
    }
]

# 2. Upgrade the referee to check for complex sequential logic
def industrial_referee(prompt, code):
    if not code.strip().endswith(";"):
        return "❌ FAIL: Syntax Error! Missing a semicolon."

    if "Flip-Flop" in prompt and "always" not in code:
        return "❌ FAIL: Logic Error! Sequential memory requires an 'always' block."

    if "Multiplexer" in prompt and "?" not in code:
        return "❌ FAIL: Logic Error! MUX requires conditional ternary logic."

    return "✅ PASS: Advanced hardware logic verified!"

# 3. Test the new factory referee against the D Flip-Flop
print("--- TESTING INDUSTRIAL REFEREE ---")
for example in advanced_dataset:
    verdict = industrial_referee(example["prompt"], example["code"])
    print(f"Component: {example['prompt']}\nVerdict: {verdict}\n")
```

```python
# 1. Create the advanced training prompt
advanced_training_prompt = """
Prompt: Design a 2-to-1 Multiplexer (MUX)
Code: assign out = sel ? b : a;

Prompt: Design a 1-bit Half Adder
Code: assign sum = a ^ b; assign carry = a & b;

Prompt: Design a D Flip-Flop memory cell
Code:"""

# 2. Let the AI generate a response using our fine-tuned creative parameters
print("The AI is attempting to design advanced sequential memory logic...")
advanced_ai_output = ai_brain(
    advanced_training_prompt,
    max_length=90,
    repetition_penalty=1.3,
    do_sample=True,
    temperature=0.3
)

# 3. Extract the generated text
advanced_ai_text = advanced_ai_output[0]['generated_text']
print("\n--- RAW AI SEQUENTIAL GENERATION ---")
print(advanced_ai_text)

# 4. Isolate the final code attempt for the Flip-Flop and test it with our industrial referee
try:
    ai_flipflop_code = advanced_ai_text.split("Design a D Flip-Flop memory cell\nCode:")[-1].strip().split("\n")[0]
    print("\n--- INDUSTRIAL REFEREE REPORT ---")
    print(f"AI Generated Code: {ai_flipflop_code}")
    print(f"Verdict: {industrial_referee('Design a D Flip-Flop memory cell', ai_flipflop_code)}")
except Exception as e:
    print("\nParsing failed. Let's check the raw output above.")
```

```python
import time

# 1. Initialize variables for our automated hardware debugger
current_prompt = advanced_training_prompt
max_attempts = 3
success = False

print("--- STARTING ADVANCED AUTONOMOUS REPAIR LOOP ---\n")

for attempt in range(1, max_attempts + 1):
    print(f"🎬 Attempt #{attempt}: Querying AI brain for Verilog sequence...")

    # Generate code based on the current context
    ai_raw_output = ai_brain(
        current_prompt,
        max_length=120,
        repetition_penalty=1.3,
        do_sample=True,
        temperature=0.4
    )[0]['generated_text']

    # Isolate the newest attempt line safely
    try:
        generated_part = ai_raw_output.split("Design a D Flip-Flop memory cell\nCode:")[-1].strip()
        ai_code_attempt = generated_part.split("\n")[0].strip()
    except Exception:
        ai_code_attempt = "invalid_syntax;"

    print(f"🤖 AI Generated Line: {ai_code_attempt}")

    # Run the attempt through our rigid industrial inspector
    verdict = industrial_referee("Design a D Flip-Flop memory cell", ai_code_attempt)
    print(f"📋 Referee Verdict: {verdict}")

    if "✅ PASS" in verdict:
        print(f"\n🎉 CRITICAL SUCCESS! Flawless sequential logic generated after {attempt} correction cycles!")
        print(f"Final Validated Verilog Code: {ai_code_attempt}")
        success = True
        break
    else:
        print("🔄 Circuit failure detected. Compiling feedback text loop...")

        # Inject programmatic correction hints directly into the next prompt context
        if "always" in verdict:
            hint = "\nSystem Feedback: FAIL! You must use an sequential 'always @(posedge clk)' block for memory elements."
        else:
            hint = "\nSystem Feedback: FAIL! Syntactically broken statement. Ensure it closes with a semicolon ';'"

        # Dynamic context reinforcement
        current_prompt += f" {ai_code_attempt}\n{hint}\nPrompt: Design a D Flip-Flop memory cell\nCode:"
        print("-" * 50)
        time.sleep(1)

if not success:
    print("\n⚠️ AI maxed out available attempts. This clearly demonstrates the boundary limits of an 82M parameter text-model processing complex real-time RTL state-machines!")
```

```python
# 1. Expand to a comprehensive industrial hardware design specification matrix
enterprise_hardware_matrix = [
    {"component": "4-bit Arithmetic Logic Unit (ALU)", "spec": "combinatorial", "prompt": "Design a 4-bit ALU supporting ADD and SUB operations.", "expected_keywords": ["assign", "sum", "sub"]},
    {"component": "8-bit Shift Register", "spec": "sequential", "prompt": "Design an 8-bit serial-in parallel-out shift register.", "expected_keywords": ["always", "posedge", "clk", "<="]},
    {"component": "3-to-8 Decoder", "spec": "combinatorial", "prompt": "Design a 3-to-8 binary decoder circuit.", "expected_keywords": ["assign", "case", ";"]},
    {"component": "Synchronous Up-Counter", "spec": "sequential", "prompt": "Design a 4-bit synchronous up-counter with reset.", "expected_keywords": ["always", "posedge", "clk", "reset"]},
    {"component": "4x4 Read-Only Memory (ROM)", "spec": "sequential", "prompt": "Design a 4x4 bit ROM array matrix.", "expected_keywords": ["always", "reg", "memory"]}
]

# 2. Advanced batch evaluation loop engine
def run_enterprise_benchmark():
    print("🚀 INITIALIZING ENTERPRISE BATCH BENCHMARK RUN...\n" + "="*60)
    passed_count = 0
    total_count = len(enterprise_hardware_matrix)

    for idx, item in enumerate(enterprise_hardware_matrix, 1):
        print(f"📦 Evaluating Component {idx}/{total_count}: {item['component']}")
        print(f"📋 Spec Class: {item['spec'].upper()}")

        # Simulate neural language generation based on the target prompt
        # (Using a controlled simulation to evaluate rule coverage across all 5 industrial components)
        if item["spec"] == "combinatorial":
            simulated_ai_code = "assign out = (opcode == 0) ? (a + b) : (a - b);"
        else:
            # Intentional missing timing block to test deep verification catch rates
            simulated_ai_code = "assign q_next = d; // missing clock boundary"

        print(f"🤖 Raw AI Synthesis Attempt: {simulated_ai_code}")

        # Verify against specific architectural keywords required for that chip class
        violations = [kw for kw in item["expected_keywords"] if kw not in simulated_ai_code]

        if not violations:
            print("Verdict: ✅ PASSED STRUCTURAL VALIDATION!")
            passed_count += 1
        else:
            print(f"Verdict: ❌ FAILED! Missing explicit structural hardware elements: {violations}")

        print("-" * 60)

    yield_rate = (passed_count / total_count) * 100
    print(f"📊 BENCHMARK SUMMARY: Compilers verified {passed_count}/{total_count} circuits.")
    print(f"📐 Silicon Success Yield Rate: {yield_rate}%")

# 3. Spin up the batch verification pipeline
run_enterprise_benchmark()
```
